```python
data["timestamp"] = pd.to_datetime(data["timestamp"])

# Sort by timestamp (useful for time-based analysis later)
data = data.sort_values("timestamp").reset_index(drop=True)

# Extract time-based features
data["hour"] = data["timestamp"].dt.hour
data["day_of_week"] = data["timestamp"].dt.dayofweek  # 0=Monday, 6=Sunday
data["day"] = data["timestamp"].dt.day
data["month"] = data["timestamp"].dt.month
```

```python
print(data[["transaction_id", "timestamp", "hour", "day_of_week", "month"]].head())
```

```python
data["upi_handle"] = data["sender_upi"].str.split("@").str[1].str.lower()

# --- count unique handles ---

handle_count = data["upi_handle"].nunique()
print(handle_count)
```
